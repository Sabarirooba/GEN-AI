```python
import pandas as pd
import numpy as np

# 1. Dataset
data = {
    'Marketing_Spend': [2000,5000,1000,7000,8000,3000,9000,4000],
    'Customer_Visits': [50,120,30,150,200,60,180,90],
    'Product_Quality': [6,8,5,9,9,7,8,6],
    'Competition_Level':[8,3,9,2,1,7,2,5],
    'Profit':           [0,1,0,1,1,0,1,1]
}
df = pd.DataFrame(data)

X = df[['Marketing_Spend','Customer_Visits','Product_Quality','Competition_Level']].values
y = df['Profit'].values

# 2. Initialize weights & bias
w = np.zeros(X.shape[1])
b = 0.0
learning_rate = 0.0001   # small learning rate (because spend values are large)
epochs = 1000

# 3. Activation function (Step function)
def activation(z):
    return np.where(z >= 0, 1, 0)

# 4. Training loop
for epoch in range(epochs):
    for i in range(len(X)):
        z = np.dot(w, X[i]) + b
        y_pred = activation(z)
        error = y[i] - y_pred

        # Update weights & bias
        w = w + learning_rate * error * X[i]
        b = b + learning_rate * error

# 5. Predictions on dataset
df['Predictions'] = activation(np.dot(X, w) + b)
print("DATAFRAME AFTER PREDICTION:\n", df)

# 6. Accuracy
correct = np.sum(df['Profit'] == df['Predictions'])
total = len(df)
accuracy = correct / total * 100
print("\nAccuracy:", accuracy, "%")

# 7. User input prediction loop
while True:
    print("\n--- USER INPUT PREDICTION ---")
    Marketing = float(input("Marketing Spend: "))
    Visits = float(input("Customer Visits: "))
    Quality = float(input("Product Quality (1-10): "))
    Competition = float(input("Competition Level (1-10): "))

    user_data = np.array([Marketing, Visits, Quality, Competition])
    z_user = np.dot(w, user_data) + b
    pred_user = activation(z_user)

    print("Prediction (Profit/Loss):", pred_user)
    if pred_user == 1:
        print("✅ Business is likely to make Profit.\n")
    else:
        print("⚠️ Business may face Loss, need strategy improvement.\n")

    choice = input("Do you want to check again? (yes/no): ").casefold()
    if choice == "no":
        print("Exiting... Goodbye!")
        break
```
