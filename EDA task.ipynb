```python
#EDA
import numpy as np
import pandas as pd

# ============================================================
# 1. CREATE SALES DATA
# ============================================================

data = {
    "Product": [
        "Laptop","Phone","Headphones","Keyboard",
        "Mouse","Monitor","Laptop","Phone",
        "Headphones","Monitor","Keyboard","Mouse"
    ],
    "Category": [
        "Electronics","Electronics","Accessories","Accessories",
        "Accessories","Electronics","Electronics","Electronics",
        "Accessories","Electronics","Accessories","Accessories"
    ],
    "Quantity": [
        2,5,10,8,15,3,1,4,12,2,6,20
    ],
    "Price": [
        55000,30000,2000,1500,800,25000,
        55000,30000,2000,25000,1500,800
    ],
    "Rating": [
        4.5,4.2,4.7,4.1,4.6,4.3,4.8,4.0,
        4.5,4.4,4.2,4.7
    ]
}

df = pd.DataFrame(data)

print("="*60)
print("        E-COMMERCE SALES INTELLIGENCE SYSTEM")
print("="*60)

print("\nOriginal Sales Data:")
print(df)

# ============================================================
# 2. UNDERSTAND THE DATA
# ============================================================

print("\n\n--- DATASET INFORMATION ---")
print("\nShape:", df.shape)
print("\nColumns:", df.columns)
print("\nData Types:\n", df.dtypes)

# ============================================================
# 3. CALCULATE REVENUE USING NUMPY
# ============================================================

df["Revenue"] = np.multiply(df["Quantity"], df["Price"])

print("\n\n--- REVENUE CALCULATION ---")
print(df[["Product","Quantity","Price","Revenue"]])

# ============================================================
# 4. BASIC NUMPY STATISTICS
# ============================================================

revenue = df["Revenue"].to_numpy()

print("\n\n--- REVENUE STATISTICS ---")
print("Total Revenue:", np.sum(revenue))
print("Average Revenue:", np.mean(revenue))
print("Highest Revenue:", np.max(revenue))
print("Lowest Revenue:", np.min(revenue))
print("Revenue Standard Deviation:", np.std(revenue))

# ============================================================
# 5. FIND THE HIGHEST REVENUE TRANSACTION
# ============================================================

highest_index = np.argmax(revenue)
print("\n\n--- HIGHEST REVENUE TRANSACTION ---")
print(df.iloc[highest_index])

# ============================================================
# 6. PRODUCT-WISE SALES
# ============================================================

product_sales = df.groupby("Product")["Quantity"].sum()
print("\n\n--- PRODUCT-WISE SALES ---")
print(product_sales)

best_selling_product = product_sales.idxmax()
best_selling_quantity = product_sales.max()
print("\nBest Selling Product:", best_selling_product)
print("Units Sold:", best_selling_quantity)

# ============================================================
# 7. PRODUCT-WISE REVENUE
# ============================================================

product_revenue = df.groupby("Product")["Revenue"].sum()
print("\n\n--- PRODUCT-WISE REVENUE ---")
print(product_revenue)

highest_revenue_product = product_revenue.idxmax()
highest_revenue_value = product_revenue.max()
print("\nHighest Revenue Product:", highest_revenue_product)
print("Revenue:", highest_revenue_value)

# ============================================================
# 8. CATEGORY-WISE REVENUE
# ============================================================

category_revenue = df.groupby("Category")["Revenue"].sum()
print("\n\n--- CATEGORY-WISE REVENUE ---")
print(category_revenue)

# ============================================================
# 9. HIGHEST RATED PRODUCTS
# ============================================================

high_rated_products = df[df["Rating"] >= 4.5]
print("\n\n--- HIGHLY RATED PRODUCTS ---")
print(high_rated_products[["Product","Rating"]])

# ============================================================
# 10. AVERAGE RATING
# ============================================================

average_rating = np.mean(df["Rating"].to_numpy())
print("\nAverage Product Rating:", round(average_rating,2))

# ============================================================
# 11. FIND UNUSUALLY HIGH REVENUE
# ============================================================

mean_revenue = np.mean(revenue)
std_revenue = np.std(revenue)
threshold = mean_revenue + std_revenue

unusual_sales = df[df["Revenue"] > threshold]
print("\n\n--- UNUSUALLY HIGH REVENUE ---")
print(unusual_sales[["Product","Quantity","Revenue"]])

# ============================================================
# 12. CREATE PERFORMANCE SCORE
# ============================================================

df["Performance_Score"] = (df["Rating"]*20 + df["Quantity"]*2)
print("\n\n--- PRODUCT PERFORMANCE SCORE ---")
print(df[["Product","Rating","Quantity","Performance_Score"]])

# ============================================================
# 13. SORT PRODUCTS BY PERFORMANCE
# ============================================================

ranked_products = df.sort_values("Performance_Score", ascending=False)
print("\n\n--- TOP PERFORMING TRANSACTIONS ---")
print(ranked_products[["Product","Performance_Score"]].head(5))

# ============================================================
# 14. NUMPY SLICING
# ============================================================

print("\n\n--- NUMPY SLICING DEMO ---")
quantity_array = df["Quantity"].to_numpy()
print("All quantities:", quantity_array)
print("First 5 quantities:", quantity_array[:5])
print("Last 5 quantities:", quantity_array[-5:])

# ============================================================
# 15. NUMPY SORTING
# ============================================================

print("\n\n--- NUMPY SORTING ---")
sorted_revenue = np.sort(revenue)
print("Revenue from lowest to highest:", sorted_revenue)

# ============================================================
# 16. NORMALIZATION DEMO
# ============================================================

print("\n\n--- NUMPY NORMALIZATION ---")
normalized_revenue = revenue / np.max(revenue)
print(normalized_revenue)

# ============================================================
# 17. FINAL BUSINESS REPORT
# ============================================================

print("\n")
print("="*60)
print("              FINAL BUSINESS REPORT")
print("="*60)
print("\nTotal Revenue: ₹", np.sum(revenue))
print("Average Transaction Revenue: ₹", round(np.mean(revenue),2))
print("Best Selling Product:", best_selling_product)
print("Highest Revenue Product:", highest_revenue_product)
print("Highest Revenue Transaction: ₹", np.max(revenue))
print("Average Rating:", round(average_rating,2))
print("Number of Highly Rated Transactions:", len(high_rated_products))
print("Unusually High Revenue Transactions:", len(unusual_sales))
print("="*60)
```

```python
report_data = {
    "Metric": [
        "Total Revenue",
        "Average Transaction Revenue",
        "Best Selling Product",
        "Highest Revenue Product",
        "Highest Revenue Transaction",
        "Average Rating",
        "Highly Rated Transactions",
        "Unusually High Transactions"
    ],
    "Value": [
        f"₹ {np.sum(revenue)}",
        round(np.mean(revenue),2),
        best_selling_product,
        highest_revenue_product,
        f"₹ {np.max(revenue)}",
        round(average_rating,2),
        len(high_rated_products),
        len(unusual_sales)
    ]
}

final_report = pd.DataFrame(report_data)
print(final_report)
```
