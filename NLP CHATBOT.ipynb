```python
import nltk
import pandas as pd
nltk.download('punkt')
df=pd.read_csv('Conversation csv.csv')
responses=dict(zip(df['question'],df['answer']))
while True:
    user_input=input('You:').lower()
    if user_input=='bye':
        print('Bot:Goodbye!')
        break

    if user_input in responses:
        print('Bot:', responses[user_input])
    else:
        print("Bot: Sorry, I don't understand.")
        
```
