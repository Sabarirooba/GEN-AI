```python
#Apply function
import pandas as pd
data={'Employ':['Riya','Aadhi','Tanish','Bharath'],
      'Department':['HR','HR','IT','Marketing'],
      'Salary':[49000,49000,51000,52000]}
df=pd.DataFrame(data)
print(df)
```

```python
#Bonus=10% of salary
df['Bonus']=df['Salary'].apply(lambda x:x*0.10)
print(df)
```

```python
import pandas as pd
data={'Name':['Ini','Sidhu','Tamil','Roja'],
      'Marks':[98,85,86,95],
      'Subject':['English','Maths','Science','Social']}
df=pd.DataFrame(data)
print(df)
```

```python
df['Result']=df['Marks'].apply(lambda x:'very good' if x>=95 else 'good')
print(df)
```
