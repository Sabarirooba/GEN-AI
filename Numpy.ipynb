```python
flat=marks.flatten()
print('flatten:',flat)
```

```python
#Ravel()
rav=marks.ravel()
print('ravel:',rav)
```

```python
#Difference check
rav[0]=100
print('Ravel after change:',rav)
print('Original after ravel change:\n',marks)
flat[1]=99
print('Flatten after change:',flat)
print('Original after flatten change:\n',marks)
```

```python
import numpy as np
marks = np.array([
    [85, 90, 78],
    [88, 76, 92],
    [90, 85, 80],
    [70, 95, 88]
])
print('total sum:',np.sum(marks))
print('mean:',np.mean(marks))
print('min:',np.min(marks))
print('max:',np.max(marks))
print('median:',np.median(marks))
print('std dev:',np.std(marks))
print('variance:',np.var(marks))
print('row-wise sum:',np.sum(marks,axis=1))
print('column-wise sum:',np.sum(marks,axis=0))

```

```python
#Linear algebra
#1.Dot product
import numpy as np
m1=([1,2,3])
m2=([4,5,6])
print('dot product:',np.dot(m1,m2))
```

```python
#2.Matrix multiplication
m3=([
    [1,2],
    [3,4]
])
m4=([
    [5,6],
    [7,8]
])
print('matrix multiplication:\n',np.matmul(m3,m4))
print('determinant:',np.linalg.det(m3))
print('inverse:',np.linalg.inv(m3))
```

```python
#Sorting
unsorted = np.array([3, 1, 4, 1, 5, 9, 2, 6])
print(np.sort(unsorted))
print(np.argsort(unsorted))
print(np.where(unsorted > 3))

```
