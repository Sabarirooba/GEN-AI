```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
import pandas as pd

# Example dataset
train_df = pd.DataFrame({
    "message": ["hi", "hello", "balance check", "loan request", "bye"],
    "category": ["greet", "greet", "balance", "loan", "exit"]
})

vectorizer = TfidfVectorizer(ngram_range=(1,2), max_features=2000)
X_train = vectorizer.fit_transform(train_df["message"])
y_train = train_df["category"]

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

print("✅ Model Training Complete")
```

```python
def predict_intent(message):
    X_test = vectorizer.transform([message])
    intent = model.predict(X_test)[0]
    confidence = max(model.predict_proba(X_test)[0])
    return intent, confidence

def chatbot(message):
    intent, confidence = predict_intent(message)
    if confidence <= 0.50:
        return gemini.generate_content(message).text
    else:
        return f"Intent: {intent}, Confidence: {round(confidence*100,2)}%"
```

```python
print("=====================================")
print(" SAVARTBANK NLP + GEMINI CHATBOT ")
print(" type 'exit' to quit")
print("=====================================")

while True:
    user_input = input("You: ")
    if user_input.lower() == "exit":
        print("SmartBank: Goodbye!")
        break
    result = chatbot(user_input)
    print("SmartBank:", result)
```
